```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
import re
```

```python
lemmatizer = WordNetLemmatizer()
def lammatize_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    words = text.split()
    lammatized_words=[lemmatizer.lemmatize(word) for word in words if word not in stopwords.words('english')]
    return " ".join(lammatized_words)
```

```python
df['title'] = df['title'].apply(lammatize_text)
df.head()
```

```python
df['text'] = df['text'].apply(lammatize_text)
df.head()
```

```python
df=pd.read_csv('Dataset.csv')
df.head()
```

```python
df['subject']=df['subject'].replace('politics','politicsNews')
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
df=df.drop('date',axis=1)
```

```python
y=df['Label']
X=df.drop(columns='Label',axis=1)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
VOCAB_SIZE = 25000
TITLE_MAX_LENGTH = 100
TEXT_MAX_LENGTH = 1500
```

```python
all_texts = pd.concat([X_train['title'], X_train['text']], ignore_index=True).astype(str)
tokenizer = Tokenizer(num_words=VOCAB_SIZE, oov_token="<OOV>")
tokenizer.fit_on_texts(all_texts)
```

```python
def process_text_with_tokenizer(text_series, tokenizer, max_length):
    """
    Takes a pandas Series of text, a fitted tokenizer, and a max length,
    and returns padded integer sequences.
    """
    # a. Convert texts to sequences of integers
    sequences = tokenizer.texts_to_sequences(text_series.astype(str))
    
    # b. Pad the sequences to a uniform length
    padded_sequences = pad_sequences(
        sequences,
        maxlen=max_length,
        padding='post',
        truncating='post'
    )
    return padded_sequences
```

```python
padded_title_sequences = process_text_with_tokenizer(X_train['title'], tokenizer, TITLE_MAX_LENGTH)
X_train['title_vectors'] = list(padded_title_sequences)
```

```python
padded_text_sequences = process_text_with_tokenizer(X_train['text'], tokenizer, TEXT_MAX_LENGTH)
X_train['text_vectors'] = list(padded_text_sequences)
```

```python
padded_title_sequences = process_text_with_tokenizer(X_test['title'], tokenizer, TITLE_MAX_LENGTH)
X_test['title_vectors'] = list(padded_title_sequences)
```

```python
padded_text_sequences = process_text_with_tokenizer(X_test['text'], tokenizer, TEXT_MAX_LENGTH)
X_test['text_vectors'] = list(padded_text_sequences)
```

```python
X_train.head()
```

```python
X_train=X_train.drop(columns=['title','text'],axis=1)
```

```python
X_test=X_test.drop(columns=['title','text'],axis=1)
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder(sparse_output=False)
```

```python
enco_train=encoder.fit_transform(X_train[['subject']])
new_column_names = encoder.get_feature_names_out(['subject'])
encoded_df = pd.DataFrame(
    enco_train,
    columns=new_column_names,
    index=X_train.index
)
X_train = pd.concat([X_train.drop('subject', axis=1), encoded_df], axis=1)
```

```python
enco_test=encoder.transform(X_test[['subject']])
new_column_names_test = encoder.get_feature_names_out(['subject'])
encoded_df = pd.DataFrame(
    enco_test,
    columns=new_column_names,
    index=X_test.index 
)
X_test = pd.concat([X_test.drop('subject', axis=1), encoded_df], axis=1)
```
